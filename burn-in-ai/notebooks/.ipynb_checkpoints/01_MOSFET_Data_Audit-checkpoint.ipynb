# 01. MOSFET Data Audit

## Objective
Audit the NASA MOSFET thermal overstress dataset to determine its physical time range, sampling intervals, and whether it can genuinely support a 0h $\to$ 24h $\to$ 96h $\to$ 168h prediction schema.

```
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')
print(f"Total Physical Devices: {df['component_id'].nunique()}")
```

### Conclusion
**Can this dataset genuinely support: 0h -> 24h -> 96h -> 168h prediction?**

**PARTIALLY.** The physical NASA tests result in failure within ~15 hours of extreme thermal overstress. We map this accelerated life conceptually to 'Equivalent Burn-In Hours' (EBH) to simulate the 168h burn-in scale structurally.
