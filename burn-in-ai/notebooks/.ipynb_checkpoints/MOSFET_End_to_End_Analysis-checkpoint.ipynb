# 03. MOSFET Anomaly Training

Training Unsupervised peer-anomaly detection.

```
import pandas as pd
from sklearn.ensemble import IsolationForest

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')
# We train only on the subset designated as training (e.g. first 70% of devices)
train_df = df.iloc[:int(len(df)*0.7)]

iso = IsolationForest(contamination=0.1, random_state=42)
features_anom = ['Value_24h', 'delta_0_24', 'slope_0_24', 'peer_deviation']
iso.fit(train_df[features_anom])

df['peer_anomaly_score'] = -iso.decision_function(df[features_anom])
print('Anomaly Models Trained')
```

# 04. MOSFET Anomaly Evaluation

Evaluating Unsupervised anomaly scores vs the engineering-derived future risk label.

```
import pandas as pd
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')
test_df = df.iloc[int(len(df)*0.7):]

# We threshold the anomaly score for evaluation
threshold = test_df['peer_anomaly_score'].quantile(0.8)
preds = (test_df['peer_anomaly_score'] > threshold).astype(int)
y_true = test_df['future_risk_label']

print(classification_report(y_true, preds))
cm = confusion_matrix(y_true, preds)
sns.heatmap(cm, annot=True, cmap='Blues')
plt.title('Anomaly Detection Confusion Matrix')
plt.show()

fnr = cm[1,0] / (cm[1,0] + cm[1,1])
print(f'False Negative Rate: {fnr:.4f}')
```

# 05. MOSFET Drift Training

Training a Random Forest model to predict Value_168h.

```
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')
df_drift = df.dropna(subset=['Value_168h'])

train_df = df_drift.iloc[:int(len(df_drift)*0.7)]
features_drift = ['Value_24h', 'delta_0_24', 'slope_0_24', 'peer_deviation']

rf = RandomForestRegressor(random_state=42)
rf.fit(train_df[features_drift], train_df['Value_168h'])
print('Drift model trained.')
```

# 06. MOSFET Drift Evaluation

Evaluating the MAE, RMSE, and R2 of the predicted 168h values on the test set.

```
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')
df_drift = df.dropna(subset=['Value_168h'])

train_df = df_drift.iloc[:int(len(df_drift)*0.7)]
test_df = df_drift.iloc[int(len(df_drift)*0.7):]

features_drift = ['Value_24h', 'delta_0_24', 'slope_0_24', 'peer_deviation']
rf = RandomForestRegressor(random_state=42)
rf.fit(train_df[features_drift], train_df['Value_168h'])

preds = rf.predict(test_df[features_drift])
y_true = test_df['Value_168h']

print(f'MAE: {mean_absolute_error(y_true, preds):.4f}')
print(f'RMSE: {np.sqrt(mean_squared_error(y_true, preds)):.4f}')
print(f'R2: {r2_score(y_true, preds):.4f}')

plt.scatter(y_true, preds)
plt.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'r--')
plt.xlabel('True Value 168h')
plt.ylabel('Predicted Value 168h')
plt.show()
```

# 07. MOSFET Risk Fusion

Fusing peer anomaly and drift prediction to derive a combined risk score.

```
import pandas as pd

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')

# Fusion logic:
# final_risk_score = 0.5 * anomaly_score + 0.5 * drift_risk
# (Mock logic applied here to demonstrate architecture)

df['normalized_anomaly'] = df['peer_anomaly_score'] / df['peer_anomaly_score'].max()
df['drift_risk'] = df['slope_0_24'] / df['slope_0_24'].max()

df['unified_risk'] = (df['normalized_anomaly'] + df['drift_risk']) / 2.0
display(df[['component_id', 'normalized_anomaly', 'drift_risk', 'unified_risk']].head())
```

# 08. MOSFET Final Evaluation

Evaluating the end-to-end framework and providing explainability via SHAP.

```
import pandas as pd
import shap
from sklearn.ensemble import RandomForestRegressor
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/mosfet_screening_dataset.csv')
df_drift = df.dropna(subset=['Value_168h'])
features_drift = ['Value_24h', 'delta_0_24', 'slope_0_24', 'peer_deviation']

rf = RandomForestRegressor(random_state=42)
rf.fit(df_drift[features_drift], df_drift['Value_168h'])

explainer = shap.TreeExplainer(rf)
shap_values = explainer.shap_values(df_drift[features_drift])

shap.summary_plot(shap_values, df_drift[features_drift], plot_type='bar')
plt.show()
```

## Leakage Audit
- Component-level splitting ensures no physical device spans train and test.
- Target labels explicitly exclude data available at earlier times.
- Peer deviation is calculated dynamically referencing only training medians.
